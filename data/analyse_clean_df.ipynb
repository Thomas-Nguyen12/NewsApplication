```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import MultiLabelBinarizer
from iterstrat.ml_stratifiers import MultilabelStratifiedShuffleSplit

def get_minority_samples(X, y, ir_threshold=1.0):
    """Get samples belonging to minority labels (below mean IR)."""
    n_samples = y.shape[0]
    label_counts = y.sum(axis=0)
    mean_ir = np.mean(label_counts)
    
    minority_labels = np.where(label_counts < mean_ir)[0]
    minority_mask = y[:, minority_labels].sum(axis=1) > 0
    
    return X[minority_mask], y[minority_mask]

def mlsmote(X, y, n_samples=500, k_neighbors=5, random_state=42):
    """
    MLSMOTE: Synthetic Minority Oversampling for Multilabel data
    
    Parameters:
        X: np.ndarray — feature matrix (numeric)
        y: np.ndarray — binary label matrix (n_samples x n_labels)
        n_samples: int — number of synthetic samples to generate
        k_neighbors: int — number of nearest neighbors to use
    
    Returns:
        X_resampled, y_resampled (originals + synthetic)
    """
    rng = np.random.default_rng(random_state)
    
    # Focus on minority samples
    X_min, y_min = get_minority_samples(X, y)
    
    if len(X_min) < k_neighbors + 1:
        print(f"Warning: only {len(X_min)} minority samples, reducing k")
        k_neighbors = len(X_min) - 1
    
    # Fit KNN on minority samples
    knn = NearestNeighbors(n_neighbors=k_neighbors + 1)  # +1 because sample is own neighbor
    knn.fit(X_min)
    
    synthetic_X = []
    synthetic_y = []
    
    for _ in range(n_samples):
        # Pick a random minority sample as reference
        idx = rng.integers(0, len(X_min))
        x_ref = X_min[idx]
        y_ref = y_min[idx]
        
        # Find its k nearest neighbors
        _, neighbor_idxs = knn.kneighbors([x_ref])
        neighbor_idxs = neighbor_idxs[0][1:]  # exclude self
        
        # Pick a random neighbor
        nn_idx = rng.choice(neighbor_idxs)
        x_neighbor = X_min[nn_idx]
        y_neighbor = y_min[nn_idx]
        
        # Interpolate features
        gap = rng.random()
        x_synthetic = x_ref + gap * (x_neighbor - x_ref)
        
        # Union of labels (standard MLSMOTE label generation)
        y_synthetic = np.logical_or(y_ref, y_neighbor).astype(int)
        
        synthetic_X.append(x_synthetic)
        synthetic_y.append(y_synthetic)
    
    X_resampled = np.vstack([X, np.array(synthetic_X)])
    y_resampled = np.vstack([y, np.array(synthetic_y)])
    
    return X_resampled, y_resampled
```

```python
vectorizer = TfidfVectorizer(max_features=500, stop_words=stop_words)
X = vectorizer.fit_transform(X['text_cleaned'].str.lower()).toarray()  # must be dense
y = y_mlb
# --- Stratified split BEFORE resampling ---
splitter = MultilabelStratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_idx, test_idx in splitter.split(X, y):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

X_train_res, y_train_res = mlsmote(X_train, y_train, n_samples=300, k_neighbors=5)

print(f"Before: {X_train.shape[0]} samples")
print(f"After:  {X_train_res.shape[0]} samples")

# --- Check label distribution improved ---
before = y_train.sum(axis=0)
after = y_train_res.sum(axis=0)
comparison = pd.DataFrame({'before': before, 'after': after}, index=mlb.classes_)
print(comparison)
```
