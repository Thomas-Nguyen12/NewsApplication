```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import MultiLabelBinarizer
from iterstrat.ml_stratifiers import MultilabelStratifiedShuffleSplit

def get_minority_samples(X, y, ir_threshold=1.0):
    """Get samples belonging to minority labels (below mean IR)."""
    n_samples = y.shape[0]
    label_counts = y.sum(axis=0)
    mean_ir = np.mean(label_counts)
    
    minority_labels = np.where(label_counts < mean_ir)[0]
    minority_mask = y[:, minority_labels].sum(axis=1) > 0
    
    return X[minority_mask], y[minority_mask]

def mlsmote(X, y, n_samples=500, k_neighbors=5, random_state=42):
    """
    MLSMOTE: Synthetic Minority Oversampling for Multilabel data
    
    Parameters:
        X: np.ndarray — feature matrix (numeric)
        y: np.ndarray — binary label matrix (n_samples x n_labels)
        n_samples: int — number of synthetic samples to generate
        k_neighbors: int — number of nearest neighbors to use
    
    Returns:
        X_resampled, y_resampled (originals + synthetic)
    """
    rng = np.random.default_rng(random_state)
    
    # Focus on minority samples
    X_min, y_min = get_minority_samples(X, y)
    
    if len(X_min) < k_neighbors + 1:
        print(f"Warning: only {len(X_min)} minority samples, reducing k")
        k_neighbors = len(X_min) - 1
    
    # Fit KNN on minority samples
    knn = NearestNeighbors(n_neighbors=k_neighbors + 1)  # +1 because sample is own neighbor
    knn.fit(X_min)
    
    synthetic_X = []
    synthetic_y = []
    
    for _ in range(n_samples):
        # Pick a random minority sample as reference
        idx = rng.integers(0, len(X_min))
        x_ref = X_min[idx]
        y_ref = y_min[idx]
        
        # Find its k nearest neighbors
        _, neighbor_idxs = knn.kneighbors([x_ref])
        neighbor_idxs = neighbor_idxs[0][1:]  # exclude self
        
        # Pick a random neighbor
        nn_idx = rng.choice(neighbor_idxs)
        x_neighbor = X_min[nn_idx]
        y_neighbor = y_min[nn_idx]
        
        # Interpolate features
        gap = rng.random()
        x_synthetic = x_ref + gap * (x_neighbor - x_ref)
        
        # Union of labels (standard MLSMOTE label generation)
        y_synthetic = np.logical_or(y_ref, y_neighbor).astype(int)
        
        synthetic_X.append(x_synthetic)
        synthetic_y.append(y_synthetic)
    
    X_resampled = np.vstack([X, np.array(synthetic_X)])
    y_resampled = np.vstack([y, np.array(synthetic_y)])
    
    return X_resampled, y_resampled
```

```python
X = cleaned_df['text_cleaned']
y = cleaned_df.topic_split.str.split(",")
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=0.2)
```

```python
Y_train = mlb.transform(Y_train)
Y_test = mlb.transform(Y_test)
```

```python
batch_size = 32
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
import numpy as np

# ── Tokenisation ──────────────────────────────────────────────────────────────
VOCAB_SIZE = 5000
MAX_LEN    = 200
EMBED_DIM  = 64

tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)

# Then proceed as normal
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)

# --------------------
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq  = tokenizer.texts_to_sequences(X_test)

X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LEN, padding="post", truncating="post")
X_test_pad  = pad_sequences(X_test_seq,  maxlen=MAX_LEN, padding="post", truncating="post")

# ── Model ─────────────────────────────────────────────────────────────────────
NUM_LABELS = Y_train.shape[1]   # inferred from your label matrix

model = Sequential([
    Embedding(input_dim=VOCAB_SIZE, output_dim=EMBED_DIM, input_length=MAX_LEN),
    Bidirectional(LSTM(128, return_sequences=True)),  # reads text both ways
    Dropout(0.3),
    LSTM(128),
    Dropout(0.3),
    Dense(128, activation="relu"),
    Dense(11, activation="sigmoid"),          # sigmoid for multi-label
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",   # correct loss for multi-label
    metrics=["accuracy"],
)

model.summary()

# ── Callbacks ─────────────────────────────────────────────────────────────────
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,               # stop if val_loss doesn't improve for 3 epochs
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,               # halve the learning rate on plateau
    patience=2,
    min_lr=1e-6,
    verbose=1
)

# ── Training ──────────────────────────────────────────────────────────────────
history = model.fit(
    X_train_pad, Y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test_pad, Y_test),
    callbacks=[early_stop, reduce_lr],
    verbose=1,
)

# ── Evaluation ────────────────────────────────────────────────────────────────
loss, accuracy = model.evaluate(X_test_pad, Y_test, verbose=0)
print(f"Test loss:     {loss:.4f}")
print(f"Test accuracy: {accuracy:.4f}")

# ── Save ──────────────────────────────────────────────────────────────────────
model.save("data/lstm_classifier.h5")
joblib.dump(tokenizer, "data/lstm_tokenizer.pkl")
print("Model and tokenizer saved.")
```
